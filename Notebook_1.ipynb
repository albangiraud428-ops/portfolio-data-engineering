```python
# Welcome to your new notebook
# Type here in the cell editor to add code!

df = spark.read.option("multiline", "true").json("abfss://Projet_Portfolio@onelake.dfs.fabric.microsoft.com/Crypto_Lakehouse.Lakehouse/Files/crypto_raw.json")
# df now is a Spark DataFrame containing JSON data from "abfss://Projet_Portfolio@onelake.dfs.fabric.microsoft.com/Crypto_Lakehouse.Lakehouse/Files/crypto_raw.json".
display(df)
```

```python
from pyspark.sql.functions import explode, col

# 1. On "éclate" (explode) la liste contenue dans la colonne 'coins' pour créer une ligne par crypto
df_coins = df.select(explode(col("coins")).alias("crypto"))

# 2. On extrait les informations utiles qui sont imbriquées dans 'crypto.item'
df_silver = df_coins.select(
    col("crypto.item.id").alias("id"),
    col("crypto.item.name").alias("nom"),
    col("crypto.item.symbol").alias("symbole"),
    col("crypto.item.price_btc").alias("prix_btc"),
    col("crypto.item.market_cap_rank").alias("rang")
)

# 3. On affiche le résultat nettoyé !
display(df_silver)
```

```python
# Sauvegarde du DataFrame en tant que table Delta dans le Lakehouse
df_silver.write.format("delta").mode("overwrite").saveAsTable("crypto_silver")
```
